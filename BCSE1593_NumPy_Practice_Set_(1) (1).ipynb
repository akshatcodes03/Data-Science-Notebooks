### Question 15 – Combining two measurements

For each month, the operations team wants to use the actual electricity consumption when it is available. If the value is 0, it should use a backup estimate.

Create two small NumPy arrays:

```text
actual = [42000, 0, 44800, 0, 47800]
backup = [41000, 43000, 44000, 45500, 47000]
```

Use `np.where()` to create a final array that:

- uses `backup` when `actual` is 0,
- otherwise uses `actual`.


````python
# Question 15
actual = np.array([42000, 0, 44800, 0, 47800])
backup = np.array([41000, 43000, 44000, 45500, 47000])
final_energy = np.where(actual == 0, backup, actual)
print("Final energy:", final_energy)
````

### Question 24 – Applying a monthly adjustment

The university wants to identify classrooms where occupancy was **below 50% at least once** during the week.

### Question 35 – Energy efficiency check

The university wants to compare electricity consumption with student visits.

For each facility:

1. Calculate total yearly electricity consumption.
2. Calculate total yearly student visits.
3. Calculate electricity consumption per student visit.
4. Display the three resulting values.
5. Identify the facility with the highest electricity consumption per visit.

````python
# Question 35
yearly_energy_by_facility = np.sum(monthly_energy, axis=0)
yearly_visits_by_facility = np.sum(monthly_visits, axis=0)
energy_per_visit = yearly_energy_by_facility / yearly_visits_by_facility
most_energy_per_visit_index = np.argmax(energy_per_visit)
print("Yearly energy:", yearly_energy_by_facility)
print("Yearly visits:", yearly_visits_by_facility)
print("Energy per student visit:", energy_per_visit)
print("Highest energy per visit:", facilities[most_energy_per_visit_index],
      "with", energy_per_visit[most_energy_per_visit_index], "kWh per visit")
````
