## Q27. Create travel_df, add np.arange(5), then replace with a partial Series.

```python
travel_df = pd.DataFrame({
    "Destination": ["Delhi", "Agra", "Jaipur", "Goa", "Manali"]
})

travel_df["Code"] = np.arange(5)
print("After np.arange(5):")
print(travel_df)

travel_codes = pd.Series(["TR101", "TR104"], index=[1, 4])
travel_df["Code"] = travel_codes

print("After assigning partial Series:")
print(travel_df)
```

## Q28. Create school_df, request missing House column, then fill it.

```python
school_df = pd.DataFrame({
    "Student": ["A", "B", "C", "D"],
    "Class": [10, 10, 11, 11],
    "Section": ["A", "B", "A", "B"]
})

school_selected_df = school_df.reindex(
    columns=["Student", "Class", "Section", "House"]
)

print("Before assigning House:")
print(school_selected_df)

school_selected_df["House"] = "Tagore"
print("After assigning common House value:")
print(school_selected_df)
```

## Q29. Add a partially indexed Series to inventory_df, then delete it.

```python
inventory_df = pd.DataFrame({
    "Item": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Quantity": [5, 20, 15, 8, 3],
    "Price": [60000, 800, 1500, 12000, 15000]
})

stock_series = pd.Series(
    ["Warehouse A", "Warehouse B"],
    index=[1, 3]
)

inventory_df["Warehouse"] = stock_series

print("After adding partially indexed Series:")
print(inventory_df)

inventory_df = inventory_df.drop(columns=["Warehouse"])

print("After deleting Warehouse:")
print(inventory_df)
```

## Q30. Complete sequence for lab_df using a copied table.

```python
import pandas as pd
from io import StringIO


data = """
Name       Age    City       Score
Nandini    19     Mathura    85
Akshat     20     Delhi      92
Aryan      19     Mumbai     78
Rudra      21     Lucknow    88
"""


lab_df = pd.read_csv(StringIO(data.strip()), sep=r"\s+")


print("--- 1. Complete DataFrame ---")
print(lab_df)


print("\n--- 2. Column Names ---")
print(lab_df.columns)


print("\n--- 3. Single Column (Name) ---")
print(lab_df["Name"])


print("\n--- 4. Column (City) ---")
print(lab_df["City"])


print("\n--- 5. Selected Columns ---")
small_lab_df = lab_df[["Name", "Age", "City"]]
print(small_lab_df)


print("\n--- 6. Added Status Column ---")
lab_df["Status"] = "Checked"
print(lab_df)


print("\n--- 7. After Deleting Status Column ---")
lab_df = lab_df.drop(columns=["Status"])
print(lab_df)
```
