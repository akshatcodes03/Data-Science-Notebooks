4. Display the three resulting values.
5. Identify the facility with the highest electricity consumption per visit.

Use vectorized NumPy operations.

This is the final step of the case study: turning raw numerical arrays into a simple decision that can help campus management.


````python
# 1. Calculate total yearly electricity consumption for each facility
total_yearly_energy = np.sum(monthly_energy, axis=0)

# 2. Calculate total yearly student visits for each facility
total_yearly_visits = np.sum(monthly_visits, axis=0)

# 3. Calculate electricity consumption per student visit
energy_per_visit = total_yearly_energy / total_yearly_visits

# 4. Display the three resulting values for each facility
for facility, energy, visit, epv in zip(facilities, total_yearly_energy, total_yearly_visits, energy_per_visit):
    print(f"{facility}: {epv:.4f} kWh/visit (Energy: {energy}, Visits: {visit})")

# 5. Identify the facility with the highest electricity consumption per visit
highest_epv_index = np.argmax(energy_per_visit)
print("\nFacility with highest consumption per visit:", facilities[highest_epv_index])
````
